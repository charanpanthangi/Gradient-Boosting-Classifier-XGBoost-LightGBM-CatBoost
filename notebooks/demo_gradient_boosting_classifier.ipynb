# Gradient Boosting Classifier Demo

This notebook demonstrates XGBoost, LightGBM, and CatBoost on the scikit-learn breast cancer dataset.

## What is boosting?
Boosting builds an ensemble of weak learners (trees) sequentially. Each learner focuses on the mistakes of the previous ones, steadily improving the model.

## Library differences
- **XGBoost:** Flexible with strong regularization options.
- **LightGBM:** Fast histogram-based splits, great for large datasets.
- **CatBoost:** Handles categorical features gracefully and works well with default settings.

```python
import pandas as pd
from app.data import load_data
from app.model import get_models, train_models
from app.preprocess import train_test_split_data
from app.evaluate import evaluate_models
from app.visualize import plot_feature_importances, plot_confusion_matrices
```

```python
# Load and split the dataset
X, y = load_data()
X_train, X_test, y_train, y_test = train_test_split_data(X, y)
X.shape, y.shape
```

```python
models = get_models()
trained_models = train_models(models, X_train, y_train)
metrics = evaluate_models(trained_models, X_test, y_test)
metrics
```

```python
# Plot feature importances and confusion matrices
feature_paths = plot_feature_importances(trained_models, list(X.columns))
confusion_paths = plot_confusion_matrices(metrics)
feature_paths, confusion_paths
```
